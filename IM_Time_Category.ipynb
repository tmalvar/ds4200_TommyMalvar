```python
# import all of the libraries
import pandas as pd
import altair as alt
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load the dataset into this file
df = pd.read_csv("Intramural_Clean.csv")

df.head()
```

#### Visualization 1: Does time influence the number of forfeits?

```python
# generate boxplot to understand time and forfeit status


# parse Time_Cleaned correctly (24-hour with seconds)
times = pd.to_datetime(df["Time_Cleaned"], format="%H:%M:%S", errors="coerce")
df["Minutes"] = times.dt.hour * 60 + times.dt.minute

# keep only rows with valid Minutes
df_plot = df[df["Minutes"].notna()]

# generate boxplot to understand time and forfeit status
chart = (
    alt.Chart(df_plot)
    .mark_boxplot(size=30)
    .encode(
        x = alt.X("Forfeit:N", title="Forfeit Status"),
        y = alt.Y(
            "Minutes:Q",
            title = "Game Time (24-hour)",
            axis = alt.Axis(
                labelExpr = (
                    "floor(datum.value/60) + ':' + "
                    "(datum.value%60 < 10 ? '0' : '') + datum.value%60"
                )
            ),
        ),
        color = alt.Color("Forfeit:N", legend = alt.Legend(title="Forfeit")),
    )
    .properties(
        width = 350,
        height = 350,
        title = "Game Time Distribution by Forfeit Status",
    )
)

chart
```

The boxplot shows the relationship between forfeit status and the times of every game. The games in orange indicated games that were forfeited, whereas the blue games were not. According to the plot, there were slightly fewer games that were forfeited before 1:30 PM, and the majority of games were forfeited after 6 PM. The majority of games played occur after 6 PM, as shown by the boxes. However, there is no direct indicator showing that time affects forfeits, as the box size of both statuses are roughly the same at the same times.
